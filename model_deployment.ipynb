```python
import glob
from ultralytics import YOLO

m = YOLO("/content/drive/MyDrive/Colab Notebooks/projet_mlops/best_nonqat.pt")
p = str(m.export(format="litert", imgsz=640, quantize=None))   # FP32
print("Export :", p)

f = p if p.endswith(".tflite") else glob.glob(f"{p}/**/*.tflite", recursive=True)[0]
import tensorflow as tf
it = tf.lite.Interpreter(model_path=f); it.allocate_tensors()
print("FICHIER :", f)
print("ENTREE  :", it.get_input_details()[0]["shape"], it.get_input_details()[0]["dtype"])
for o in it.get_output_details():
    print("SORTIE  :", o["shape"], o["dtype"])
```
